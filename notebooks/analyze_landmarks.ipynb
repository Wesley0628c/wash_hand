# Hand Landmark & Feature Distribution Analysis
This notebook analyzes and visualizes the hand landmarks, palm normals, and inter-hand distances across different wash-hand actions.

```python
import numpy as np
import matplotlib.pyplot as plt
import glob
import os
import sys
sys.path.append('..')

from src.rule_classifier import LABELS, LABEL_SHORT_ZH
from src.train import load_dataset_from_disk, generate_synthetic_dataset
```

```python
# Load data or generate synthetic sample
X, y, groups = load_dataset_from_disk('../data/processed')
if len(X) == 0:
    print("No recorded data found, generating synthetic distribution...")
    X, y, groups = generate_synthetic_dataset(num_persons=5, clips_per_action=10)

print(f"Loaded dataset shape: {X.shape}, Labels: {len(y)}")
```

```python
# Plot feature distribution comparison across actions
plt.figure(figsize=(12, 6))
for label_idx in range(len(LABELS)):
    idx = np.where(y == label_idx)[0]
    if len(idx) > 0:
        sample_feats = X[idx].mean(axis=(0, 1))
        plt.plot(sample_feats[:30], label=f"{LABELS[label_idx]} ({LABEL_SHORT_ZH[LABELS[label_idx]]})")

plt.title("Mean Feature Profiles Across 7-Step Wash Actions")
plt.xlabel("Feature Index")
plt.ylabel("Normalized Value")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
